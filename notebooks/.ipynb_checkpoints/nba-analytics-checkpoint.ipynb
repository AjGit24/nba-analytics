```python
import pandas as pd
from sqlalchemy import create_engine
import os

# Load CSVs
regular = pd.read_csv('../data/2023-2024 NBA Player Stats - Regular.csv', sep = ";", encoding='latin-1')
playoffs = pd.read_csv('../data/2023-2024 NBA Player Stats - Playoffs.csv', sep = ";", encoding='latin-1')

# Create SQLite Database (Using absolute path)
db_path = os.path.abspath('../data/nba.db')
engine = create_engine(f'sqlite:///{db_path}')

# Load both dataframes into the database
regular.to_sql('regular_season', engine, if_exists='replace', index=False)
playoffs.to_sql('playoffs', engine, if_exists='replace', index=False)

print("Tables successfully loaded")
```

```python
# Top 10 scorers in the regular season 
query = """
SELECT Player, Tm, PTS, AST, TRB
FROM regular_season
WHERE Tm != 'TOT'
ORDER BY PTS DESC
LIMIT 10
"""

top_scorers = pd.read_sql(query, engine)
print(top_scorers)
```

```python
# Top 10 most efficient scorers (Minimum 20 games and 15 minutes played)
query2 = """
SELECT Player, Tm, ROUND("FG%",3) as FG_PCT, ROUND ("eFG%", 3) as eFG_PCT, PTS
FROM regular_season
WHERE Tm != 'TOT' AND G >= 20 AND MP >= 15
ORDER BY "eFG%" DESC
LIMIT 10
"""
efficient_scorers = pd.read_sql(query2, engine)
print(efficient_scorers)
```

```python
# Top 10 assisters with low turnover rates assist to turnover ratio)
query3 = """
SELECT Player, Tm, AST, TOV, ROUND(AST/NULLIF(TOV, 0), 2) as AST_TO_ratio
FROM regular_season
WHERE Tm != 'TOT' AND G >= 20
ORDER BY AST_TO_ratio DESC
LIMIT 10
"""
playmakers = pd.read_sql(query3, engine)
print(playmakers)
```

```python
# Regular Season vs. Playoffs Comparison
query4 = """
SELECT r.Player, r.Tm, r.PTS as Regular_PTS, p.PTS as Playoff_PTS, ROUND(p.PTS - r.PTS, 1) as PTS_DIFFERENCE
FROM regular_season r 
JOIN playoffs p on r.player = p.player
where r.TM != 'TOT' AND p.TM != 'TOT' AND r.G >= 20 AND p.G >= 5
ORDER BY PTS_DIFFERENCE DESC
LIMIT 10
"""
clutch_players = pd.read_sql(query4, engine)
print(clutch_players)

```

```python
queries = {
    "top_scorers.sql": """ 
    SELECT Player, Tm, PTS, AST, TRB
    FROM regular_season
    WHERE Tm != 'TOT'
    ORDER BY PTS DESC
    LIMIT 10
    """,

    "efficient_scorers.sql": """
    SELECT Player, Tm, ROUND("FG%",3) as FG_PCT, ROUND ("eFG%", 3) as eFG_PCT, PTS
    FROM regular_season
    WHERE Tm != 'TOT' AND G >= 20 AND MP >= 15
    ORDER BY "eFG%" DESC
    LIMIT 10
    """,

    "playmakers.sql": """
    SELECT Player, Tm, AST, TOV, ROUND(AST/NULLIF(TOV, 0), 2) as AST_TO_ratio
    FROM regular_season
    WHERE Tm != 'TOT' AND G >= 20
    ORDER BY AST_TO_ratio DESC
    LIMIT 10
    """,

    "clutch_players.sql": """
    SELECT r.Player, r.Tm, r.PTS as Regular_PTS, p.PTS as Playoff_PTS, ROUND(p.PTS - r.PTS, 1) as PTS_DIFFERENCE
    FROM regular_season r 
    JOIN playoffs p on r.player = p.player
    where r.TM != 'TOT' AND p.TM != 'TOT' AND r.G >= 20 AND p.G >= 5
    ORDER BY PTS_DIFFERENCE DESC
    LIMIT 10
    """
}

import os
for filename, query in queries.items():
    with open(f'../sql/{filename}', 'w') as f:
        f.write(query)

print("SQL file saved!")
    
```

```python
top_scorers.to_csv('../data/top_scorers.csv', index=False)
efficient_scorers.to_csv('../data/efficient_scorers.csv', index=False)
playmakers.to_csv('../data/playmakers.csv', index=False)
clutch_players.to_csv('../data/clutch_players.csv', index=False)
```
